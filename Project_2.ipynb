```python
import pandas as pd
import matplotlib.pyplot as plt
import cartopy.crs as ccrs

# Загрузите датасет
df = pd.read_csv('spicam_obs.csv')

# Фильтруем точки, где есть лед CO2 и H2O
co2_ice = df[df['co2ice_index'] > 2]
h2o_ice = df[df['h2oice_index'] > 0.15]

fig = plt.figure(figsize=(12, 8))
ax = plt.axes(projection=ccrs.Mollweide())

ax.set_title('Лёд на Марсе: CO2 (красный), H2O (синий)', fontsize=16)

# CO2 лед — красные точки
plt.scatter(co2_ice['lon'], co2_ice['lat'], color='red', label='CO2 лёд', s=8, alpha=0.6, transform=ccrs.PlateCarree())

# H2O лед — синие точки
plt.scatter(h2o_ice['lon'], h2o_ice['lat'], color='blue', label='H2O лёд', s=8, alpha=0.6, transform=ccrs.PlateCarree())

ax.set_xlabel('Longitude')
ax.set_ylabel('Latitude')
ax.gridlines(draw_labels=True)
plt.legend()
plt.show()
```
